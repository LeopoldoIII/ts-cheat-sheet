### 📖 1. Code Sample: Generic Containers & Utility Types

Generics allow functions and interfaces to capture types as variables. Utility types transform existing types without manual duplication:

````typescript
// 1. Generic Wrapper
interface ApiResponse<T> {
    data: T;
    status: number;
    timestamp: number;
}

interface User {
    id: number;
    username: string;
    email: string;
}

// 2. Utility Types
type UserPreview = Pick<User, "id" | "username">; // Only id and username
type UserUpdatePayload = Partial<User>;           // All fields optional

const response: ApiResponse<UserPreview> = {
    data: { id: 1, username: "octocat" },
    status: 200,
    timestamp: Date.now(),
};

console.log("User response:", response.data.username);
````
