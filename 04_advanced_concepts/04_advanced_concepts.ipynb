# 04. Advanced Concepts: Generics and Utility Types

TypeScript shines when you use its advanced type system to create reusable and highly flexible components.

## 1. Generics
Generics allow you to pass types as variables (`<T>`) so your functions and classes can work with any data type while still remaining type-safe.

```typescript
// <T> captures the type provided by the user
function identity<T>(arg: T): T {
    return arg;
}

// Explicitly setting the type:
const num = identity<number>(42);

// Relying on type inference (TS knows it's a string):
const str = identity("Hello Generics");

console.log("Number:", num);
console.log("String:", str);
```

## 2. Generic Classes
You can also create classes that wrap or manage generic types.

```typescript
class GenericBox<T> {
    private content: T;

    constructor(value: T) {
        this.content = value;
    }

    getContent(): T {
        return this.content;
    }
}

const stringBox = new GenericBox<string>("Secret Message");
console.log("Box content:", stringBox.getContent());
```
