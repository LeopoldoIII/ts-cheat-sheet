# 07. Best Practices: SOLID Principles

Applying standard software engineering practices like SOLID to TypeScript.

## Single Responsibility Principle (SRP)
A class should have one, and only one, reason to change.

```typescript
// ❌ BAD: This class handles both user logic AND database saving
class UserBad {
    constructor(public name: string) {}

    public saveToDatabase() {
        console.log(`Saving ${this.name} to the database...`);
    }
}

// ✅ GOOD: Separate the responsibilities
class User {
    constructor(public name: string) {}
}

class UserRepository {
    public save(user: User) {
        console.log(`Saving ${user.name} to the database...`);
    }
}

const user = new User("Alice");
const repo = new UserRepository();
repo.save(user);
```
