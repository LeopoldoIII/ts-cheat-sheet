### 📖 1. Code Sample: Dependency Inversion & Guard Clauses

High-level business logic should depend on abstractions (interfaces), not concrete implementations:

````typescript
// 1. Abstraction
interface Logger {
    log(message: string): void;
}

// 2. Concrete implementation
class ConsoleLogger implements Logger {
    log(message: string): void {
        console.log(`[LOG]: ${message}`);
    }
}

// 3. Service depending on the abstraction
class UserService {
    constructor(private logger: Logger) {}

    registerUser(email: string): void {
        // Guard clause: Early exit
        if (!email.includes("@")) {
            throw new Error("Invalid email address");
        }
        this.logger.log(`Registered user: ${email}`);
    }
}

const service = new UserService(new ConsoleLogger());
service.registerUser("alice@example.com");
````
