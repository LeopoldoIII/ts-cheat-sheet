# 03. Functional Programming in JS/TS

JavaScript treats functions as **first-class citizens**, meaning you can pass them around just like any other variable. This makes functional programming (FP) patterns very common.

## 1. Higher-Order Functions: map, filter, reduce

These are the bread and butter of array manipulation in modern JS/TS.

```typescript
const numbers: number[] = [1, 2, 3, 4, 5];

// map: Transforms each element into something else
const doubled = numbers.map((num) => num * 2);
console.log("Doubled:", doubled);

// filter: Returns a new array with elements that pass the condition
const evens = numbers.filter((num) => num % 2 === 0);
console.log("Evens:", evens);

// reduce: Accumulates all elements into a single value
// The '0' is the initial value of 'acc'
const sum = numbers.reduce((acc, curr) => acc + curr, 0);
console.log("Sum:", sum);
```

## 2. Promises and Async/Await

Functional programming in JS heavily relies on asynchronous operations to avoid blocking the main thread.

```typescript
// Simulating an asynchronous operation (like fetching from a DB)
const fetchFakeData = (): Promise<string> => {
    return new Promise((resolve) => {
        setTimeout(() => {
            resolve("Data loaded successfully!");
        }, 1000);
    });
};

console.log("Fetching data...");

// Using top-level await (supported natively in Deno/Notebooks!)
const result = await fetchFakeData();
console.log(result);
```
