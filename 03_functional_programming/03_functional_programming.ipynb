Master functional programming idioms in TypeScript. In modern JavaScript/TypeScript, functions are first-class citizens. You will learn to use pure functions and Higher-Order Functions (`map`, `filter`, `reduce`) to transform data declaratively without mutating original inputs.

### 📖 1. Code Sample: Transforming Collections with HOFs

Instead of imperative `for` loops with accumulators, functional approaches chain array methods to create clean, readable data pipelines:

````typescript
interface Product {
    id: number;
    name: string;
    price: number;
    inStock: boolean;
}

const inventory: Product[] = [
    { id: 1, name: "Mechanical Keyboard", price: 120, inStock: true },
    { id: 2, name: "Wireless Mouse", price: 50, inStock: false },
    { id: 3, name: "Gaming Monitor", price: 300, inStock: true },
];

// Filter in-stock items, extract prices, and sum them up immutably
const totalInStockValue = inventory
    .filter((item) => item.inStock)
    .map((item) => item.price)
    .reduce((accumulator, price) => accumulator + price, 0);

console.log("Total In-Stock Value:", totalInStockValue); // 420
````

function calculateOrderSummary(items: CartItem[]): OrderSummary {
    return items.reduce<OrderSummary>(
        (acc, item) => {
            const lineTotal = item.unitPrice * item.quantity;
            const lineTax = item.isTaxExempt ? 0 : lineTotal * 0.10;
            return {
                subtotal: acc.subtotal + lineTotal,
                totalTax: acc.totalTax + lineTax,
                grandTotal: acc.grandTotal + lineTotal + lineTax,
                itemCount: acc.itemCount + item.quantity,
            };
        },
        { subtotal: 0, totalTax: 0, grandTotal: 0, itemCount: 0 }
    );
}
