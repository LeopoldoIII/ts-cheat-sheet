# 06. Integrations and APIs

Here we demonstrate how to fetch data from a REST API and use TypeScript Interfaces to type the JSON response.

## 1. Fetching Data with strict types
When you call `.json()`, TS assumes it's `any`. By casting it to an Interface, we get safety and autocomplete.

```typescript
// Define the shape of the data we expect from the API
interface UserAPIResponse {
    id: number;
    name: string;
    email: string;
    username: string;
}

try {
    const response = await fetch("https://jsonplaceholder.typicode.com/users/1");
    if (!response.ok) {
        throw new Error(`HTTP error! status: ${response.status}`);
    }
    
    // Cast the generic JSON response to our defined Interface
    const data = (await response.json()) as UserAPIResponse;
    
    // TS now gives us autocomplete and safety!
    console.log(`User name: ${data.name}, Email: ${data.email}`);
    
    // This would fail in TypeScript:
    // console.log(data.invalidField);
} catch (error) {
    console.error("Fetch failed:", error);
}
```
