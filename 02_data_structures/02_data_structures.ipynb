```typescript
function countWordFrequencies(words: string[]): Map<string, number> {
    const frequencies = new Map<string, number>();
    for (const word of words) {
        const currentCount = frequencies.get(word) ?? 0;
        frequencies.set(word, currentCount + 1);
    }
    return frequencies;
}
```
