# 02. Data Structures: Arrays, Objects, Maps, and Sets

Let's see how data structures work natively, and how TypeScript adds safety via Interfaces and explicit types.

## 1. Arrays and Tuples
In JS, arrays can hold mixed types. In TS, we usually lock them down to a specific type, or use a `Tuple` if we know the exact length and types.

```typescript
// TS: Strictly typed array
const numbersArray: number[] = [1, 2, 3, 4, 5];
// numbersArray.push("six"); // This would fail
console.log("Numbers:", numbersArray);

// TS: Tuple (fixed length and types, similar to Python tuples)
const userRecord: [string, number, boolean] = ["Alice", 28, true];
console.log("Tuple:", userRecord);
```

## 2. Objects and Interfaces
Objects in JS are dynamic (dictionaries). In TS, we use `Interfaces` or `Types` to define their exact shape, giving us autocomplete and preventing typos.

```typescript
interface User {
    name: string;
    age: number;
    isActive: boolean;
    // The '?' makes this property optional
    role?: string;
}

const myUser: User = {
    name: "John Doe",
    age: 30,
    isActive: true
};

// myUser.invalidProp = "Test"; // Fails in TS!
myUser.role = "Admin"; // Allowed because it's optional in the Interface

console.log("User:", myUser);
```

## 3. Maps and Sets
Modern JS provides `Map` and `Set`. TS adds generics to these to ensure we only insert the correct types.

```typescript
// A Set of strings (no duplicates allowed)
const uniqueNames = new Set<string>();
uniqueNames.add("Alice");
uniqueNames.add("Bob");
uniqueNames.add("Alice"); // Ignored, already exists

console.log("Set size:", uniqueNames.size);

// A Map with string keys and number values
const userAges = new Map<string, number>();
userAges.set("Alice", 28);
userAges.set("Bob", 34);

console.log("Alice's age:", userAges.get("Alice"));
```
