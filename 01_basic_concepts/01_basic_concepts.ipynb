# 01. Basic Concepts: Variables and Control Flow

In this notebook, we will explore the core differences between dynamic typing in JavaScript and static typing in TypeScript.

## 1. Dynamic Typing (JavaScript Style)
In standard JavaScript, variables do not have a fixed type. You can assign a string, and later assign a number to the exact same variable.

```typescript
// JS Style Declaration
let myVariable = "Hello World";
console.log("String:", myVariable);

// Re-assigning to a different type is valid in JS (but generally a bad practice)
myVariable = 42;
console.log("Number:", myVariable);
```

## 2. Static Typing (TypeScript Style)
In TypeScript, you can explicitly define the type. If you try to re-assign it to a different type later, the TS compiler (and this Deno kernel) will throw an error.

```typescript
// TS Style Declaration
let myTypedVariable: string = "Hello TypeScript";
console.log(myTypedVariable);

// Uncommenting the next line will cause a TypeError in TypeScript!
// myTypedVariable = 42; 
```

## 3. Control Flow
Control flow like `if`, `else if`, and `else` works exactly the same in both JS and TS. The only difference is typing the parameters in the function definition.

```typescript
// We explicitly define `num` as a number, and specify that the function returns a string.
const checkNumber = (num: number): string => {
    if (num > 0) {
        return "Positive";
    } else if (num < 0) {
        return "Negative";
    } else {
        return "Zero";
    }
};

console.log("Check Number 42:", checkNumber(42));
console.log("Check Number -5:", checkNumber(-5));
```
