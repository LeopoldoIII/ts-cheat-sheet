# 09. Object-Oriented Programming (OOP)

TypeScript adds powerful OOP features to JS, such as Interfaces, access modifiers (`private`, `public`, `protected`), and more.

## Classes and Interfaces
Interfaces define a contract that classes must follow.

```typescript
interface IAnimal {
    makeSound(): void;
}

class Animal implements IAnimal {
    // Access modifiers (public by default, can be private or protected)
    protected name: string;

    constructor(name: string) {
        this.name = name;
    }

    public makeSound(): void {
        console.log("Generic animal sound");
    }
}

class Dog extends Animal {
    private breed: string;

    constructor(name: string, breed: string) {
        super(name);
        this.breed = breed;
    }

    public makeSound(): void {
        console.log("Woof! Woof!");
    }

    public getBreed(): string {
        return this.breed;
    }
}

const myDog = new Dog("Buddy", "Golden Retriever");
console.log(`My dog is a ${myDog.getBreed()}`);
myDog.makeSound();

// Error: Property 'name' is protected
// console.log(myDog.name); 

// Error: Property 'breed' is private
// console.log(myDog.breed);
```
